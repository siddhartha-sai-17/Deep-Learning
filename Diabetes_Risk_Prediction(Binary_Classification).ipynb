 Nerual Network Architecture

2 Inputs -> 3 Hidden Neurons(sigmoid) -> Output(sigmoid)

Output = 1 -> High Risk Output = 0 -> Low Risk  

```python
import numpy as np

np.random.seed(0)

X = np.array([
    [120,85,150,95] ,#Glucose
    [25,30,35,22]  #BMI
])

#Normalize Values
X = X/200

#Output : 1=High risk , 0=Low Risk
Y = np.array([[1,0,1,0]])

print("X shape:", X.shape)
print("Y shape:",Y.shape)
```

```python
# Sigmoid function
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
# Initialize Parameters

# Hidden Layers(3 neurons,2 input)
W1 = np.random.randn(3,2)*0.1
b1 = np.zeros((3,1))

#Output Layer(1 neuron,3 hidden inputs)
w2 = np.random.randn(1,3)*0.1
b2 = np.zeros((1,1))
```

```python
#Forward Propagation

#Hidden Layer
z1 = np.dot(W1,X) + b1
a1 = sigmoid(z1)

#Output Layer
z2 = np.dot(w2,a1) + b2
a2 = sigmoid(z2)

print("Predicted Output :")
print(a2)
```

```python
# Loss Function

# Binary Cross Entrophy
m = Y.shape[1]
loss = -np.sum(Y*np.log(a2) + (1-Y)*np.log(1-a2))/m
print("Loss:",loss)
```

```python
# Back propagation

# Output Layer Gradient
dz2 = a2 - Y
dw2 = np.dot(dz2,a1.T)/m
db2 = np.sum(dz2,axis=1,keepdims=True)/m

```

```python
# Hidden Layer gradient
# Sigmoid derviate
da1 = np.dot(w2.T,dz2)
dz1 = da1*a1*(1-a1)

dw1 = np.dot(dz1,X.T)/m
db1 = np.sum(dz1,axis=1,keepdims=True)/m
```

```python
# Update the Parameters
learning_rate = 0.5
W1 = W1 - learning_rate*dw1
b1 = b1 - learning_rate*db1
w2 = w2 - learning_rate*dw2
b2 = b2 - learning_rate*db2
```

```python
# Complete Trainig Loop
epoch = 5000
for i in range(epoch):

  #Forward
  z1 = np.dot(W1,X) + b1
  a1 = sigmoid(z1)
  z2 = np.dot(w2,a1) + b2
  a2 = sigmoid(z2)

  #loss
  loss = -np.sum(Y*np.log(a2+1e-8) + (1-Y)*np.log(1-a2+1e-8))/m

  #Backward
  dz2 = a2 - Y
  dw2 = np.dot(dz2,a1.T)/m
  db2 = np.sum(dz2,axis=1,keepdims=True)/m

  dA1 = np.dot(w2.T,dz2)
  dz1 = dA1*a1*(1-a1)

  dw1 = np.dot(dz1,X.T)/m
  db1 = np.sum(dz1,axis=1,keepdims=True)/m

  #Update
  W1 = W1 - learning_rate*dw1
  b1 = b1 - learning_rate*db1
  w2 = w2 - learning_rate*dw2
  b2 = b2 - learning_rate*db2

  if i%100== 0:
    print("Epoch:",i,"Loss:",loss)
```
