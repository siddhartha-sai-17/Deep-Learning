```python
import numpy as np
```

```python
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]
              ])
y = np.array([0,1,1,0])
X,y
```

```python
w1 = 1
w2 = 1
b1 = -0.5
w3 = 1
w4 = 1
b2 = -1.5
w5 = 1
w6 = -2
b3 =-0.5
```

```python
def step_function(z):
  return 1 if z>=0 else 0
```

```python
for i in range(len(X)):
    h1 = step_function(w1*X[i][0] + w2*X[i][1] + b1)
    h2 = step_function(w3*X[i][0] + w4*X[i][1] + b2)

    y_pred = step_function(w5*h1 + w6*h2 + b3)

    print(f"Input: {X[i]}, Predicted: {y_pred}, Actual: {y[i]}")

print("\nFinal Weights:")
print(w1, w2, w3, w4, w5, w6)

print("Final Biases:")
print(b1, b2, b3)
```
