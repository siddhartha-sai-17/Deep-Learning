```python
import numpy as np
```

```python
X = np.array([[0,0],
              [0,1],
              [1,0],
              [1,1]
              ])
y = np.array([0,1,1,1])
X,y
```

```python
w = [0,0]
b = 0
epochs = 10
learning_rate = 0.1
print("Initial Weight:",w)
print("Initial Bias:",b)
```

```python
def step_function(z):
  return 1 if z>=0 else 0
```

```python

for i in range(len(X)):
        # 1. Weighted sum (dot product)
        z = np.dot(w, X[i]) + b

        # 2. Activation
        y_pred = step_function(z)

        # 3. Error
        error = y[i] - y_pred

        # 4. Update weights
        w = w + learning_rate * error * X[i]

        # 5. Update bias
        b = b + learning_rate * error

        print(f"Input: {X[i]}, Predicted: {y_pred}, Actual: {y[i]}, w: {w}, b: {b}")

print("\nFinal Weights:", w)
print("Final Bias:", b)
```
