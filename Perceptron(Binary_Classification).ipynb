***Perceptron  (Binary Classification)***

```python
import numpy as np
```

```python
#step-1 Training data
```

```python
# Training data
X = np.array([1,2,3,4]) # Study hours
y = np.array([0,0,1,1]) # Fail(0)/Pass(1)
X,y
```

```python
# step -2 : Initialize the parameters
```

```python
w = 0.0
b = 0.0
epochs = 10
learning_rate = 1
print("Initial Weight:",w)
print("Initial Bias:",b)
```

**Define Activation Function (Step Function)**

```python
def step_function(z):
  return 1 if z>=0 else 0
```

**Training Loop**

```python
for epoch in range(epochs):
    print(f"\nEpoch {epoch+1}")

    for i in range(len(X)):
        # 1. Compute weighted sum
        z = w * X[i] + b

        # 2. Apply activation function
        y_pred = step_function(z)

        # 3. Compute the error
        error = y[i] - y_pred

        # 4. Update the weight
        w = w + learning_rate * error * X[i]

        # 5. Update the bias
        b = b + learning_rate * error

        print(f"Input: {X[i]}, Predicted: {y_pred}, Actual: {y[i]}, w: {w}, b: {b}")
```

```python
print("\n Final Weight",w)
print("Final Bias",b)
```

**Testing the model**

```python
test_val = 2.9
z = w * test_val + b
prediction = step_function(z)

print(f"For Study hours: {test_val}, Prediction: {prediction}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# 1. Training Data
# -----------------------------
X = np.array([1,2,3,4])     # Study hours
y = np.array([0,0,1,1])      # 0 = Fail, 1 = Pass

# -----------------------------
# 2. Perceptron Parameters
# (Use your trained values here)
# -----------------------------
w = 1
b = -3

# -----------------------------
# 3. Decision Boundary
# w*x + b = 0  =>  x = -b / w
# -----------------------------
decision_boundary = -b / w

# -----------------------------
# 4. Plot
# -----------------------------
plt.figure()

# Plot data points
plt.scatter(X[y == 0], y[y == 0], label="Fail (0)")
plt.scatter(X[y == 1], y[y == 1], label="Pass (1)")

# Plot decision boundary
plt.axvline(x=decision_boundary, linestyle="--", label="Decision Boundary")

# Labels
plt.xlabel("Study Hours")
plt.ylabel("Class")
plt.title("Perceptron Decision Boundary")
plt.legend()

plt.show()
```
